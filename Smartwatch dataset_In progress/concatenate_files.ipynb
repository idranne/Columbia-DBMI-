```python
#Trying to put everything together
import os

subject_data = []

#Looping over the rows 
for i, row in metadata.iterrows():
    subject_id = row['id']
    label = row['label']
print(subject_id)

    # Paths
    mov_file = f"/users/imbahndu/Desktop/Columbia DBM/Smartwatch dataset/preprocessed/movement/{subject_id}_ml.bin"
    quest_idx = metadata[metadata['id'] == subject_id].index[0]
    quest_file = f"../preprocessed/questionnaire/{quest_idx+1:03d}_ml.bin"

    if not os.path.exists(mov_file) or not os.path.exists(quest_file):
        print(f"Missing file for subject {subject_id}")
        continue

    # Load data
    try:
        mov_data = np.fromfile(mov_file, dtype=np.float32).reshape(-1, 1)
        quest_data = np.fromfile(quest_file, dtype=np.float32)

        # Option 1: Concatenate questionnaire data as flat features
        quest_data_expanded = np.tile(quest_data, (mov_data.shape[0], 1))  # Repeat per time-step
        combined_data = np.hstack((mov_data, quest_data_expanded))

        # Save to list (or write to file)
        subject_data.append({
            'subject_id': subject_id,
            'features': combined_data,
            'label': label
        })
    except Exception as e:
        print(f"Error processing subject {subject_id}: {e}")
```
