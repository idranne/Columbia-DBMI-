```python
import json

def check_duplicates(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        try:
            json.load(f)
            print(f"{file_path} is valid.")
        except json.JSONDecodeError as e:
            print(f"{file_path} has JSON issues: {e}")

check_duplicates("/users/imbahndu/Desktop/Columbia DBM/Acccelometer_Gyrometer/preprocessed/movement.ipynb")
```

```python
import nbformat

path = "/users/imbahndu/Desktop/Columbia DBM/Acccelometer_Gyrometer/preprocessed/movement.ipynb"
with open(path, "r", encoding="utf-8") as f:
    nb = nbformat.read(f, as_version=4)

for cell in nb.cells:
    if cell.cell_type == "code":
        cell.execution_count = None
        cell.outputs = []

with open(path, "w", encoding="utf-8") as f:
    nbformat.write(nb, f)
```
