```python
import os
import re

def clean_conflict_markers(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        content = f.read()

    # Remove Git conflict markers
    cleaned = re.sub(r"<<<<<<< HEAD\n", "", content)
    cleaned = re.sub(r"=======\n", "", cleaned)
    cleaned = re.sub(r">>>>>>> [a-f0-9]+\n", "", cleaned)

    with open(file_path, "w", encoding="utf-8") as f:
        f.write(cleaned)

    print(f"Cleaned: {file_path}")

# Walk through your project directory
root_dir = "/users/imbahndu/Desktop/Columbia DBM"
for subdir, _, files in os.walk(root_dir):
    for file in files:
        if file.endswith(".ipynb"):
            full_path = os.path.join(subdir, file)
            clean_conflict_markers(full_path)
```
