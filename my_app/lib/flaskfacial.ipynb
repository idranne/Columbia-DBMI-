```python
#Importing dependencies
from flask import Flask, request, jsonify
import joblib
import numpy as np
from PIL import Image
import io
```

```python
#Directory and model load 
app = Flask(__name__)

model - joblib.load("C:\Users\User\Columbia-DBMI-\PFED5\my_cnn_model.h5")
```

```python
#Preprocess the image for CNN
def preprocess_image(image_bytes):

    #Ensuring RGB
    img = Image.open(io.BytesIO(image_bytes)).convert("RGB")

    #Size match to match the training sample
    img = img.resize((123, 128))

    #Normalize as training set
    img_array = np.array(img)/255.0

    img_array = np.expand_dims(img_array, axis = 0)

    return img_array

```

```python
'''----------------------------PREDICTION ROUTE--------------'''

@app.route("/predict_facial", methods = ["POST"])

#function of predict 
def predict_facial():
    #request variable from the .dart page 
    if "file" not in request.files:
        return jsonify({"Error": "No file uploaded"}), 400
    
    #if the file is found,
    file = request.files["file"]
    #read the file 
    image_bytes = file.read()
    #call the predict method
    img_array = preprocess_image(image_bytes)
```

```python
#Predict with CNN

#return probabilities
prediction = model.predict(img_array)

predicted_class = int(np.argmax(prediction, axis = 1)[0])

#Defining boundaries of low and high UPDRS
low_UPDRS = np.arange(0, 2, 0.1).tolist()
high_UPRDS = np.arange(2, 4, 0.1).tolist()

if predicted_class in low_UPDRS:
    return jsonify({"Your predicted UPRDS score is ", predicted_class, "and you are not as risk." : "Confidence interval is", float(np.max(prediction))})

if predicted_class in high_UPRRS:
    return jsonify({"Your predicted UPRDS score is", predicted_class, "and you are at risk." : "Confidence interval is" , float(np.max(prediction))})
```
